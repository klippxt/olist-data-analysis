## Задание 1

```python
import pandas as pd
import numpy as np

df = pd.read_csv('olist_order_items_dataset.csv')

prices = df['price'].values
freights = df['freight_value'].values

def calculate_slow(prices, freights):
  prices_list = list(prices)
  freights_list = list(freights)
  results = []
  for p, f in zip(prices_list, freights_list):
    if p > 500:
      discount = 0.15
    elif p > 100:
      discount = 0.10
    else:
      discount = 0.0
    final_price = (p * (1 - discount)) + f
    results.append(final_price)
  return results

test_results = calculate_slow(prices[:5], freights[:5])
print('res', test_results)

def calculate_fast(prices, freights):
  conditions = [
      prices > 500,
      prices > 100
  ]
  choices = [
      0.15,
      0.10
  ]
  discounts_array = np.select(conditions, choices, default=0.0)
  final_prices = (prices * (1 - discounts_array)) +  freights
  return final_prices

test_results_fast = calculate_fast(prices[:5], freights[:5])
print('fast res', test_results_fast)
print('совпадение результата', np.allclose(calculate_slow(prices, freights), calculate_fast(prices, freights)))
```

```python
t_slow = %timeit -o calculate_slow(prices, freights)
```

```python
t_fast = %timeit -o calculate_fast(prices, freights)
```

```python
print(f'ускорение: {t_slow.average / t_fast.average:.1f} раз')
```

## Задание 4

```python
feature_cols = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
features_df = products_df.dropna(subset = feature_cols).head(1000)
X = features_df[feature_cols].values

X_norm = (X - X.mean(axis = 0)) / X.std(axis = 0)
print(X_norm.shape)
print(X_norm.mean(axis = 0).round(2), X_norm.std(axis = 0).round(2))
```

```python
selected_idx = 0
selected = X_norm[selected_idx]

similarities = (X_norm @ selected) / (np.linalg.norm(X_norm, axis = 1) * np.linalg.norm(selected))

top_idx = np.argsort(similarities)[::-1][1:6]
print('выбранный товар:', features_df['product_id'].values[selected_idx])
print(features_df.iloc[top_idx][['product_id'] + feature_cols].assign(similarity = similarities[top_idx].round(4)))
```
